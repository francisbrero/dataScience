```python
import requests
import json

# Define the API endpoint and model identifier
api_url = "http://192.168.4.72:1234/v1/completions"

# Function to interact with the model
def query_model(prompt):
    headers = {
        "Content-Type": "application/json"
    }
    data = {
        "prompt": prompt,
        "max_tokens": 50000
    }
    response = requests.post(api_url, headers=headers, data=json.dumps(data))
    if response.status_code == 200:
        return response.json()
    else:
        return {"error": response.status_code, "message": response.text}

# Example usage
prompt = """You are a product manager at MadKudu working on building a demo for [Outreach.io](http://outreach.io/) to showcase our integration and justify a strategic partnership.
This demo is intended for an engineering leader at Outreach. They are considering building something themselves and have already integrated firmographics data from a vendor.

The main messages you want to convey:

1. Enterprise companies want to leverage their internal data in Sales copilots and agents in order to power Outreach for sellers.
2. MadKudu’s unique strength resides in unlocking value from internal data (data from the customer’s system which could be 1st party as well as 3rd party data they’ve bought) without a heavy lift from the RevOps team
3. MadKudu’s secret sauce resides in it’s semantic layer AI. We have an Agent called Nia which will analyze the customer’s data in order to map it to our standard semantic layer

The demo flow show start with a day-in-the-life of a seller who logs into Outreach, sees a list of prioritized accounts by MadKudu. Clicking on 3 different accounts should illustrate the following use-case:

- Reactivated Closed-opp: this account had a closed lost opportunity but recently someone from the company has been engaging with your content. This reactivation warrants reaching out to this person who’ve been surfaced by MadKudu
    
    While looking at this account we show that MadKudu has prepared an account brief for the rep and this brief uses internal as well as external data.
    

We should mention there are many other use cases and we group accounts into those as sales plays. Each play has a recommended outreach sequence to make it easy for the rep to know what to use. We can showcase 2 playbooks: expansion with companies hitting their usage limit, Linkedin content engagement from executives.

When looking at the LinkedIn content we can showcase or discuss MadKudu Linkedin Copilot which allows to write messages leveraging internal data to increase personalization.

We conclude this section by that Flor Macias (SDR at Lucid) increased her number of meetings booked from 30 to 90 in a month thanks to MadKudu

From here we should switch to the “admin” demo where we discuss how data is shown as mapped and intelligible in Copilot for reps but this is because MadKudu has a layer of mapping.

We should showcase a customer mapping where complicated events are mapped into 1 event. We should talk about the fact there are some industry-standard events every company should have certain events: viewed pricing page, requested demo, unsubscribe. While each company will track them differently, our agent identifies them and maps them. Similarly, you need to understand what defines a “customer” vs a “prospect”… This is our secret sauce, this semantic layer is critical to being able to leverage internal data and MadKudu has developed an agentic system called Nia which automated the mapping of customer data to the semantic layer. This reduces greatly the load for RevOps and makes onboarding a lot easier. Finally, Nia can suggest and make edits on behalf of RevOps thus making it easier to manage your data layer.

Please put together a demo flow and talk track based on the above. The demo should be less than 7 minutes."""
response = query_model(prompt)
answer = response["choices"][0]["text"].strip()
print(answer)
print(response)
```
